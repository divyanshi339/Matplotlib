```python
import matplotlib.pyplot as plt

expenses = [2500, 1500, 3000, 2000]
categories = ['Rent', 'Groceries', 'Utilities', 'Entertainment']
plt.pie(expenses, labels=categories, autopct='%1.1f%%', colors=["red","skyblue","green","yellow"])
plt.title("monthly expenses")
plt.savefig("pie_fig.png", dpi=300 , bbox_inches="tight")
plt.show()
```
