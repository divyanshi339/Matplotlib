```python
import matplotlib.pyplot as plt
studies_hour =[1,2,3,4,5,6,7,8]
scores =[50,55,60,65,70,75,80,80]
plt.scatter(studies_hour , scores , color="green", marker="s",label="student data")
plt.xlabel("study hour")
plt.ylabel("exam score")
plt.title("Relationship between student scores and time")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt 

plt.scatter([1,2,3], [50, 55, 60], color="red", marker="s", label="group A")
plt.scatter([1, 2, 3], [45, 50, 42], color="yellow", marker="s", label="group B")
plt.xlabel("study hour")
plt.ylabel("exam score")
plt.title("Relationship between group A and group B of student scores and time ")
plt.grid(True)
plt.legend()
plt.show()
```
