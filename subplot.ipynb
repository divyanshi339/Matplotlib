```python
import matplotlib.pyplot as plt
# first method
plt.subplot(1 , 2 , 1)
plt.plot([1,2,3,4],[10,20,15,25])
plt.title("line bar")

plt.subplot(1,2,2)
plt.bar(["A","B","C","D"],[10,20,30,15])
plt.title("bar graph")
plt.show()
```

```python
import matplotlib.pyplot as plt

#second method 

x = [1,2,3,4]
y=[10,20,15,25]

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].plot(x, y)
ax[0].set_title("line graph")

ax[1].bar(x, y)
ax[1].set_title("bar graph")
fig.suptitle("comparsion of line bar and bar graph")
plt.tight_layout() # for automatic adjustment

plt.show()

```
