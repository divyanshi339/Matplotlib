```python
import matplotlib.pyplot as plt

students = ['Alice', 'Bob', 'Charlie', 'David', 'Eva']
marks = [85, 92, 70, 90, 50]

plt.bar(students, marks , color= "orange" , edgecolor= "black" , width = 0.4)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Student Marks')
plt.legend(loc='upper right',labels=["marks of students"] , fontsize=7)
plt.show()
```
